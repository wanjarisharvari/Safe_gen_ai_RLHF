```python
import gc
import torch

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", DEVICE)
```

```python
import torch, pandas as pd, numpy as np
from transformers import AutoTokenizer, AutoModelForCausalLM
from rouge_score import rouge_scorer
from sentence_transformers import SentenceTransformer, util
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
import nltk
nltk.download("punkt", quiet=True)

test_df = pd.read_csv("/kaggle/input/datasets/sharvaiwanjari/reward-model/preference_test.csv")
prompts     = test_df["Question"].tolist()
more_pref   = test_df["More_Prefered"].tolist()
less_pref   = test_df["Less_Prefered"].tolist()


print(f"✅ Test suite loaded: {len(prompts)} samples")

rouge  = rouge_scorer.RougeScorer(["rougeL"], use_stemmer=True)
smodel = SentenceTransformer("all-MiniLM-L6-v2")
smooth = SmoothingFunction().method1

def bleu(ref, hyp):
    ref_tok = ref.split()
    hyp_tok = hyp.split()
    if len(hyp_tok) == 0:
        return 0.0
    return sentence_bleu([ref_tok], hyp_tok, smoothing_function=smooth)

def rougeL(ref, hyp):
    return rouge.score(ref, hyp)["rougeL"].fmeasure

def sem_sim(ref, hyp):
    e1 = smodel.encode(ref,  convert_to_tensor=True)
    e2 = smodel.encode(hyp,  convert_to_tensor=True)
    return util.cos_sim(e1, e2).item()

def generate_responses(model_repo, prompts, max_new_tokens=64, is_peft=False):
    print(f"  Loading {model_repo} ...")
    tok = AutoTokenizer.from_pretrained(model_repo)
    tok.pad_token = tok.eos_token

    if is_peft:
        from peft import PeftModel
        base = AutoModelForCausalLM.from_pretrained(
            "gpt2-medium", torch_dtype=torch.float16
        ).to(DEVICE)
        model = PeftModel.from_pretrained(base, model_repo).to(DEVICE)
    else:
        model = AutoModelForCausalLM.from_pretrained(
            model_repo, torch_dtype=torch.float16
        ).to(DEVICE)

    model.eval()
    responses = []
    for prompt in prompts:
        enc = tok(prompt, return_tensors="pt",
                  truncation=True, max_length=128).to(DEVICE)
        with torch.no_grad():
            out = model.generate(
                **enc,
                max_new_tokens=max_new_tokens,
                do_sample=True,
                temperature=0.7,
                pad_token_id=tok.eos_token_id,
                repetition_penalty=1.3,
            )
        resp = tok.decode(out[0][enc["input_ids"].shape[1]:],
                          skip_special_tokens=True)
        responses.append(resp if resp.strip() else "[no response]")

    del model
    gc.collect(); torch.cuda.empty_cache()
    return responses

# Evaluate responses against more_preferred
def evaluate(responses, references):
    bleu_scores  = [bleu(r, h)   for r, h in zip(references, responses)]
    rouge_scores = [rougeL(r, h) for r, h in zip(references, responses)]
    sim_scores   = [sem_sim(r, h) for r, h in zip(references, responses)]
    return {
        "BLEU":    round(np.mean(bleu_scores),  4),
        "ROUGE-L": round(np.mean(rouge_scores), 4),
        "Sem-Sim": round(np.mean(sim_scores),   4),
    }

# Set is_peft=True for LoRA/QLoRA models (they need base model + adapter)
MODELS = {
    #"Base-GPT2" : ("gpt2-medium", False),
    "PPO-Full"   : ("mysteriousgirl/ppo-gpt2-full",   False),
    #"PPO-Prefix" : ("mysteriousgirl/ppo-gpt2-prefix", False),
    "PPO-LoRA"   : ("mysteriousgirl/ppo-gpt2-lora",   True),
    "PPO-QLoRA"  : ("mysteriousgirl/ppo-gpt2-qlora",  True),
    "DPO-Full"   : ("mysteriousgirl/dpo-gpt2-full",   False),
    #"DPO-Prefix" : ("mysteriousgirl/dpo-gpt2-prefix", False),
    "DPO-LoRA"   : ("mysteriousgirl/dpo-gpt2-lora",   True),
    "DPO-QLoRA"  : ("mysteriousgirl/dpo-gpt2-qlora",  True),
}


print("\n Scoring reference columns (baseline)...")
more_vs_less = evaluate(less_pref, more_pref)
print(f"  more_preferred vs less_preferred: {more_vs_less}")

# ── Run evaluation for all models ─────────────────────
results = {}
for name, (repo, is_peft) in MODELS.items():
    print(f"\n🔄 Evaluating {name} ...")
    try:
        responses = generate_responses(repo, prompts, is_peft=is_peft)
        scores    = evaluate(responses, more_pref)
        results[name] = scores
        print(f"  ✅ {name}: {scores}")
    except Exception as e:
        print(f"  {name} failed: {e}")
        results[name] = {"BLEU": None, "ROUGE-L": None, "Sem-Sim": None}

# Print results table 
print("\n" + "=" * 65)
print("TASK C — EVALUATION RESULTS (vs more_preferred)")
print("=" * 65)
print(f"{'Model':<14} | {'BLEU':>8} | {'ROUGE-L':>8} | {'Sem-Sim':>8}")
print("-" * 65)
for name, scores in results.items():
    b  = f"{scores['BLEU']:.4f}"   if scores["BLEU"]    is not None else "  ERROR"
    r  = f"{scores['ROUGE-L']:.4f}" if scores["ROUGE-L"] is not None else "  ERROR"
    s  = f"{scores['Sem-Sim']:.4f}" if scores["Sem-Sim"] is not None else "  ERROR"
    print(f"{name:<14} | {b:>8} | {r:>8} | {s:>8}")
print("=" * 65)


results_df = pd.DataFrame(results).T.reset_index()
results_df.columns = ["Model", "BLEU", "ROUGE-L", "Sem-Sim"]
results_df.to_csv("task_c_results.csv", index=False)
print("\n Results saved to task_c_results.csv")

import matplotlib.pyplot as plt

metrics = ["BLEU", "ROUGE-L", "Sem-Sim"]
x       = np.arange(len(results))
width   = 0.25
names   = list(results.keys())

fig, ax = plt.subplots(figsize=(14, 6))
for i, metric in enumerate(metrics):
    vals = [results[n][metric] or 0 for n in names]
    ax.bar(x + i * width, vals, width, label=metric)

ax.set_xticks(x + width)
ax.set_xticklabels(names, rotation=25, ha="right")
ax.set_ylabel("Score")
ax.set_title("Task C — PPO vs DPO Performance Comparison")
ax.legend()
plt.tight_layout()
plt.savefig("task_c_comparison.png", dpi=150)
plt.show()
print("✅ Plot saved as task_c_comparison.png")
```
